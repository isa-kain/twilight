```python
# Useful little helper functions

def sep_arcsec_to_au(theta, d=100.):
    '''
    theta :: angular separation (arcseconds)
    d     :: distance to target (pc)
    '''
    
    sep = ((theta * u.arcsec).to_value(u.radian) * (d * u.pc)).to_value(u.au)
    
    return sep

def sep_au_to_arcsec(sep, d=100.):
    '''
    sep   :: physical separation (au)
    d     :: distance to target (pc)
    '''
    
    theta = ((sep * u.au).to(u.pc) / (d * u.pc) * u.radian).to_value(u.arcsec)
    
    return theta


def contrast_mag2flux(deltamag):
    
    fluxratio = 10**(deltamag / -2.5)
    return fluxratio

def contrast_flux2mag(fluxratio):
    deltamag = -2.5 * np.log10(fluxratio)
    return deltamag


def Mearth2Mjup(Me):
    Mj = (Me * u.M_earth).to_value(u.M_jup)
    return Mj

def Mjup2Mearth(Mj):
    Me = (Mj * u.M_jup).to_value(u.M_earth)
    return Me

def Rearth2Rjup(Re):
    Rj = (Re * u.R_earth).to_value(u.R_jup)
    return Rj

def Rjup2Rearth(Rj):
    Re = (Rj * u.R_jup).to_value(u.R_earth)
    return Re



# sep_arcsec_to_au(0.2)

contrast_flux2mag(1e-3)
contrast_mag2flux(12.5)
```

## Twilight TOI target pool

```python
# Read in list of TOIs from ExoFOP database.
# Parameters:
# • dec > -30
# • Confirmed planet, known planet, or planet candidate
# • No magnitude cutoffs, assuming TESS stars are bright
# • Stellar companion at separation <= 1.5 arcsec


toi_list = pd.read_csv(f'{toipath}/exofop_dec-30_starsep1500mas_2026aug03.csv', 
                       skiprows=2, on_bad_lines='warn');

print('Unique TOIs:', len(toi_list['TOI'].unique()))
display(toi_list.head())
```

```python
############################################
# Make bool columns: Observed by NIRC2 / has spectra
############################################

# For each (non-unique) instance of each unique TOI: 
# 1) does it have spectra? 2) has it been imaged by NIRC2?
has_spectra = np.full(len(toi_list), False)
has_nirc2 = np.full(len(toi_list), False)

for i, toi in enumerate(np.unique(toi_list['TOI'])):
    ix_spc = (toi_list['TOI']==toi) & (toi_list['Spectroscopy Instrument'].notnull())
    ix_nrc = (toi_list['TOI']==toi) & (toi_list['Imaging Instrument'].str.contains('NIRC2'))
    
#     print(np.sum(ix_spc), np.sum(ix_nrc))
    
    if np.sum(ix_spc) > 0.:
        has_spectra[i] = True
        
    if np.sum(ix_nrc) > 0.:
        has_nirc2[i] = True

print('Number of non-unique TOIs with spectra:', np.sum(has_spectra))
print('Number of non-unique TOIs observed with NIRC2:', np.sum(has_nirc2))


############################################
# Record deepest contrast & finest resolution for imaging
############################################

best_contrast = np.full(len(toi_list), 0.0)

for i, toi in enumerate(np.unique(toi_list['TOI'])):
    
    ix = (toi_list['TOI']==toi)
    bc = np.nanmax( toi_list[ix]['Imaging Contrast magnitude'] )
    best_contrast[i] = bc

print('Number of non-unique TOIs with contrast estimates:', np.sum(best_contrast > 0.))


# Check that these two methods are the same -- yep
np.all(~toi_list['Spectroscopy Instrument'].notnull()==toi_list['Spectroscopy Instrument'].isna())



############################################
# Make version of the list with new columns & without duplicate TOIs
############################################


# Assign new columns to dataframe
toi_list['Spectra?'] = has_spectra
toi_list['NIRC2?'] = has_nirc2
toi_list['Best imaging contrast'] = best_contrast


# Drop duplicates, just keep first
toi_list_nodup = toi_list.drop_duplicates(subset='TOI')

# Drop spectroscopy columns (inaccurate b.c. of duplicate removal)
toi_list_nodup.drop(labels=['Spectroscopy Instrument', 'Spectroscopy Spectral Resolution', 'Spectroscopy Wavelength Coverage'], axis='columns', inplace=True)

# Drop imaging columns (inaccurate b.c. of duplicate removal)
toi_list_nodup.drop(labels=['Imaging Instrument', 'Imaging Contrast magnitude', 'Imaging Contrast separation'], axis='columns', inplace=True)

print('Num unique TOIs:', len(toi_list_nodup))
display(toi_list_nodup.head())
```

```python
############################################
# Make txt starlists
############################################

# Make txt file with all unique TOI numbers: [TOI#### RA DEC Hmag]
toi_list_stripped = toi_list_nodup[['TOI', 'RA', 'Dec']] # , 'H mag'
new_TOI_col = ['TOI'+str(toi_list_stripped['TOI'].values[i]) for i in range(len(toi_list_stripped))]
toi_list_stripped['TOI'] = new_TOI_col

# Save
np.savetxt(f'{toipath}/toi_target_list.txt', toi_list_stripped.values, fmt='%s', delimiter=',')



# Make another list with Hmags included
toi_list_stripped = toi_list_nodup[['TOI', 'RA', 'H mag', 'Dec']]
new_TOI_col = ['TOI'+str(toi_list_stripped['TOI'].values[i]) for i in range(len(toi_list_stripped))]
toi_list_stripped['TOI'] = new_TOI_col
np.savetxt(f'{toipath}/toi_target_list_Hmag.txt', toi_list_stripped.values, fmt='%s', delimiter=',')


# Make another list with RA cuts
# FIXME TBD
```
